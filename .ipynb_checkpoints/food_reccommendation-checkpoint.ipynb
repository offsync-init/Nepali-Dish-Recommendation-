```python
import pandas as pd
import numpy as np
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.stem import PorterStemmer
```

```python
dataset = pd.read_csv('nepali_food_500_dishes.csv')
dataset.drop(columns = 'ID' , inplace = True)
dataset.sample(3)
```

```python
dataset.isna().sum()
```

```python
from nltk.corpus import stopwords
import re
# StopWords initialisation
stop_words =stopwords.words('english')
#Stemmer 
stemmer = PorterStemmer()

# DataCleaner
def Clean_Description(text):
    text = re.sub(r'[^\w\s]', '', text)
    text = " ".join([stemmer.stem(words.lower()) for words in nltk.word_tokenize(text) if words not in stop_words and len(words) > 2])
    return text
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

Vectoriser = TfidfVectorizer(
    preprocessor=Clean_Description,
    ngram_range=(1,2),
    max_features=100
)
```

```python
vector_space = Vectoriser.fit_transform(dataset['Description'])
```

```python
dataset[dataset['Item Name'] == 'Dal']['Description']
```

```python
from sklearn.metrics.pairwise import cosine_similarity
def Recommend_Food(user_query, top_n=5):
    item = dataset[dataset['Item Name'] == user_query]
    if item.empty:
        return "Item not found"
    
    item_description = item['Description'].values[0]  # extract string
    
    vectorised_description = Vectoriser.transform([item_description])
    
    similarities = cosine_similarity(vectorised_description, vector_space)
    
    similar_indices = similarities[0].argsort()[::-1][1:top_n+1]
    return dataset.iloc[similar_indices][['Item Name', 'Description']]
    # return item_description
```

```python
recommended_food=Recommend_Food('Sukuti')
recommended_food
```

```python
import joblib
joblib.dump(vector_space , 'vector_space.pkl')
joblib.dump(Vectoriser , 'vectoriser.pkl')
dataset.to_csv('Dataset.csv')
```
